# Custom Questions

## Question 1: How many male passengers over 40 in 2nd class survived?

```python
q1 = df.loc[
    (df["Sex"] == "male") &
    (df["Age"] > 40) &
    (df["Pclass"] == 2) &
    (df["Survived"] == 1)
]

print("Number of survivors matching criteria:", len(q1))
display(q1[["Name", "Age", "Pclass", "Survived"]])
```

```python
q1_total = df.loc[
    (df["Sex"] == "male") &
    (df["Age"] > 40) &
    (df["Pclass"] == 2)
]

print("Total male passengers over 40 in 2nd class:", len(q1_total))
print("Survived:", len(q1))
```

## Question 2: What was the average fare paid by survivors who embarked from Southampton (S)?

```python
q2 = df.loc[
    (df["Survived"] == 1) &
    (df["Embarked"] == "S"),
    "Fare"
]

print("Number of passengers matching:", len(q2))
print("Average fare:", round(q2.mean(), 2))
```

## Findings

**Q1:** Only 2 of the 21 male passengers ove 40 in 2nd class survived, so older men in 2nd class had a low survival rate

**Q2:** Survivors who embarked from Southamptonpaid an average fare of about 39.55, which is higher than the overall median fareof 14.45 from describe().

**Note:** 177 passengers have issing Age, so they are left out of the age filter in Q1.
